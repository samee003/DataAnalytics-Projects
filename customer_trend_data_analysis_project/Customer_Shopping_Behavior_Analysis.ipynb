```python
# Loading the dataset using pandas
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# Summary statistics using .describe()
df.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
# create a new column named age_group using qcut(Quantile cut)
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels) # Create four equal-sized groups based on the number of people, not the age range.
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14, #Once every 14 days/two weeks
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14, #This is ambiguous. It can mean:Once every two weeks (The same as fortnightly) OR Twice a week
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
#Checking how many people used discount with promo code
df[['discount_applied','promo_code_used']].head(10)
```

```python
#Checking if all the people who used discounts using promo code

(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
print(df['discount_applied'].value_counts())
```

```python
# OPTIMIZATION NOTE:
# If 'discount_applied' is stored as a BOOLEAN or BIT (0/1), 
# comparing integers is faster than string matching ('Yes').
# Update the WHERE clause to `discount_applied = 1` if applicable.

# Create the mapping dictionary
mapping = {'Yes': 1, 'No': 0}

# Apply it to the column
df['discount_applied'] = df['discount_applied'].map(mapping)
```
